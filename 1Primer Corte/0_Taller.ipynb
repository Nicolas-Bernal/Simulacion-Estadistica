### Ejercicio 1.

Generar $n=2000$ observaciones de las variables aleatorias independientes

$$
X_1 \sim Gamma(\alpha,1), \qquad X_2 \sim Gamma(\beta,1),
$$
con
$$
\alpha=3\ \qquad y \qquad \beta=5\
$$

Definir

$$
X=\frac{X_1}{X_1+X_2}
$$

Representar gráficamente la variable $X$ y verificar que sigue una distribución

$$
X \sim Beta(\alpha,\beta)=Beta(3,5)
$$

Finalmente, realizar una prueba de Kolmogorov-Smirnov con un nivel de significancia del \(5\%\), para determinar si los datos generados son compatibles con una distribución $Beta(3,5)$.

```r
n_sim=2000
alfa=3
beta=5

X1=rgamma(n_sim, shape=alfa, rate=1)
X2=rgamma(n_sim, shape=beta, rate=1)

X=X1/(X1+X2)

par(mfrow=c(2,2))

hist(X, freq=FALSE, breaks="FD", main="Histograma de X", xlab="X", ylab="Densidad",col="cyan")

curve(dbeta(x, shape1=alfa, shape2=beta), add=TRUE, lwd=3)

qqplot(qbeta(ppoints(n_sim), alfa, beta), X, main="Gráfico Q-Q", xlab="Cuantiles teóricos Beta", ylab="Cuantiles de X")
abline(0, 1, lwd=2, col=3)

boxplot(X, horizontal=TRUE, main="Diagrama de caja de X", xlab="X")

acf(X, main="Función de autocorrelación de X")

## Prueba de Kolmogorov-Smirnov

ks.test(x,rbeta(n_sim,alfa,beta))
```

#### Observaciones

A partir de las simulaciones realizadas se observa que:

- El histograma de la variable $X$ presenta una forma similar a la densidad teórica de una distribución $Beta(3,5)$, lo que indica un buen ajuste entre los datos simulados y la distribución propuesta.

- En el gráfico Q-Q se observa que la mayoría de los puntos se encuentran próximos a la línea de referencia, lo cual indica una correspondencia adecuada entre los cuantiles de los datos simulados y los cuantiles teóricos de la distribución $Beta(3,5)$.

- El diagrama de caja muestra que los datos se encuentran principalmente concentrados entre valores aproximados de $0.2$ y $0.5$, con algunos valores extremos hacia la parte superior, comportamiento compatible con la distribución $Beta(3,5)$.

- La función de autocorrelación presenta valores cercanos a cero para los diferentes rezagos, excepto en el rezago 0. Esto indica que las observaciones generadas no presentan una dependencia significativa entre sí, por lo que se comportan aproximadamente como observaciones independientes.

- Finalmente, la prueba de Kolmogorov-Smirnov produjo un estadístico $D=0.025$ y un valor $p=0.5596$. Como $p>0.05$, no se rechaza la hipótesis nula $H_0$. Por lo tanto, no se encuentra evidencia estadística suficiente para afirmar que los datos difieren de una distribución $Beta(3,5)$.

En conclusión, los resultados gráficos y la prueba de Kolmogorov-Smirnov son consistentes con que la variable generada

$$
X=\frac{X_1}{X_1+X_2}
$$

es compatible con una distribución $Beta(3,5)$.

### Ejercicio 2

Generar $n=2000$ observaciones de una variable aleatoria $X$ que siga una distribución Cauchy estándar:

$$
X\sim Cauchy(0,1)
$$

cuya función de densidad está dada por:

$f_X(x)=\frac{1}{\pi(1+x^2)},\qquad x\in\mathbb{R}$

Utilizando el método de inversión, obtener la expresión que permita generar la variable $X$ a partir de una variable aleatoria uniforme $U\sim U(0,1)$.

Representar gráficamente la variable generada y verificar que los datos presentan un comportamiento compatible con una distribución $Cauchy(0,1)$. Utilizar graficos para ello.

Finalmente, realizar una prueba de Kolmogorov-Smirnov con un nivel de significancia del $5$%, con el propósito de determinar si los datos generados son compatibles con una distribución $Cauchy(0,1)$.

La función de densidad de una distribución Cauchy estándar es:

$$f_X(x)=\frac{1}{\pi(1+x^2)}$$

Para obtener la función de distribución acumulada se integra la función de densidad desde $-\infty$ hasta $x$:

$$F_X(x)=\int_{-\infty}^{x}\frac{1}{\pi(1+t^2)}dt$$

Como:

$$\int\frac{1}{1+t^2}dt=\arctan(t)$$

se obtiene:

$$F_X(x)=\frac{1}{\pi}\arctan(x)+\frac{1}{2}$$

De acuerdo con el método de inversión, se toma una variable aleatoria uniforme:

$$U\sim U(0,1)$$

y se establece:

$$U=F_X(X)$$

Por lo tanto:

$$U=\frac{1}{\pi}\arctan(X)+\frac{1}{2}$$

Despejando $X$:

$$U-\frac{1}{2}=\frac{1}{\pi}\arctan(X)$$

$$\pi\left(U-\frac{1}{2}\right)=\arctan(X)$$

$$X=\tan\left[\pi\left(U-\frac{1}{2}\right)\right]$$

Por lo tanto, la expresión utilizada para generar las observaciones es:

$$\boxed{X=\tan\left[\pi\left(U-\frac{1}{2}\right)\right]}$$

```r
n_sim = 2000

U = runif(n_sim, min=0, max=1)

X = tan(pi*(U-0.5))

par(mfrow=c(2,2))

lim = quantile(X, c(0.01, 0.99))

hist(X, freq=FALSE, breaks=40,xlim=lim,main="Histograma de X",xlab="X", ylab="Densidad")

curve(dcauchy(x, location=0, scale=1),from=lim[1], to=lim[2],add=TRUE, lwd=3)

p = seq(0.01, 0.99, length.out=500)

q_teoricos = qcauchy(p, location=0, scale=1)
q_muestrales = quantile(X, p)

plot(q_teoricos, q_muestrales,main="Gráfico Q-Q",xlab="Cuantiles teóricos Cauchy",ylab="Cuantiles de X",pch=19)

abline(0, 1, lwd=2, col=3)

boxplot(X,horizontal=TRUE,outline=FALSE,main="Diagrama de caja de X",xlab="X")

acf(X,lag.max=30,main="Función de autocorrelación de X")

summary(X)

quantile(X, c(0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99))

ks.test(X, "pcauchy", location=0, scale=1)
```

#### Observaciones

- A partir de las simulaciones realizadas se observa que el histograma de la variable $X$ presenta un comportamiento compatible con la densidad teórica de una distribución $Cauchy(0,1)$. La distribución presenta simetría alrededor de cero y colas pesadas, por lo que pueden aparecer valores extremos.

- En el gráfico Q-Q, los puntos presentan una tendencia aproximadamente lineal respecto a la línea de referencia, lo que indica una correspondencia entre los cuantiles de los datos simulados y los cuantiles teóricos de la distribución $Cauchy(0,1)$.

- El diagrama de caja muestra la presencia de valores extremos, comportamiento esperado debido a las colas pesadas de la distribución Cauchy.

- La función de autocorrelación presenta valores cercanos a cero para los diferentes rezagos, excepto en el rezago $0$. Esto es consistente con la generación independiente de las observaciones.

- Finalmente, la prueba de Kolmogorov-Smirnov permite comparar los datos generados con una distribución $Cauchy(0,1)$.

Las hipótesis son:

$H_0:X\sim Cauchy(0,1)$

$H_1:X\not\sim Cauchy(0,1)$

con un nivel de significancia de:

$\alpha=0.05$

El valor obtenido cumple $p>0.05$, no se rechaza $H_0$. Por lo tanto, no existe evidencia estadística suficiente para afirmar que los datos generados difieren de una distribución $Cauchy(0,1)$.

### Ejercicio 3
Generar $n=3000$ observaciones de una variable aleatoria $X$ que siga una distribución exponencial con parámetro $\lambda$:

$$X\sim Exponencial(\lambda)$$

cuya función de densidad está dada por:

$$f_X(x)=\lambda e^{-\lambda x},\qquad x\geq0$$

Utilizando el método de inversión, obtener una expresión que permita generar la variable $X$ a partir de una variable aleatoria $U\sim U(0,1)$.

Representar gráficamente los datos generados mediante un histograma y compararlos con la densidad teórica. Finalmente, realizar una prueba de Kolmogorov-Smirnov con un nivel de significancia del $5$%, para determinar si los datos generados son compatibles con una distribución $Exponencial(\lambda)$.

La función de distribución acumulada se obtiene integrando la densidad:

$$F_X(x)=\int_0^x \lambda e^{-\lambda t}dt$$

Integrando:

$$F_X(x)=\left[-e^{-\lambda t}\right]_0^x$$

Por lo tanto:

$$F_X(x)=1-e^{-\lambda x}$$

Tomamos:

$U=F_X(X)$

donde:

$U\sim U(0,1)$

Entonces:

$$U=1-e^{-\lambda X}$$

$$e^{-\lambda X}=1-U$$

$$-\lambda X=\ln(1-U)$$

$$X=\frac{\ln(1-U)}{-\lambda}$$

o equivalentemente:

$$\boxed{X=-\frac{\ln(1-U)}{\lambda}}$$

```r
genereexp = function(nsim, lambda) {

  U = runif(nsim)
  X = -log(U)/lambda

  return(X)
}

## Para lambda=5

X = genereexp(nsim=3000, lambda=5)

par(mfrow=c(2,2))

hist(X, freq=FALSE, breaks="FD",main="Histograma de X", xlab="X", ylab="Densidad")

curve(dexp(x, rate=5),add=TRUE, lwd=3)

qqplot(qexp(ppoints(length(X)), rate=5), X, main="Gráfico Q-Q",xlab="Cuantiles teóricos Exponencial", ylab="Cuantiles de X")

abline(0, 1, lwd=2, col=3)

boxplot(X,horizontal=TRUE, main="Diagrama de caja de X",xlab="X")

acf(X,main="Función de autocorrelación de X")

ks.test(X, "pexp", rate=5)
```

####Observaciones
 - A partir de las simulaciones realizadas se observa que el histograma de la variable $X$ presenta una forma decreciente, con una mayor concentración de observaciones cerca de cero y una disminución progresiva de los valores, comportamiento característico de una distribución $Exponencial(5)$.

- La curva de densidad teórica se ajusta aproximadamente al comportamiento observado en el histograma, lo que indica una correspondencia entre los datos simulados y la distribución propuesta.

- En el gráfico Q-Q, los puntos presentan una tendencia aproximadamente lineal respecto a la línea de referencia, lo que indica una correspondencia entre los cuantiles de los datos simulados y los cuantiles teóricos de la distribución $Exponencial(5)$.

- El diagrama de caja muestra una distribución asimétrica hacia la derecha, característica de la distribución exponencial, junto con algunos valores extremos positivos.

- La función de autocorrelación presenta valores cercanos a cero para los diferentes rezagos, excepto en el rezago $0$, lo que es consistente con la generación independiente de las observaciones.

- Finalmente, la prueba de Kolmogorov-Smirnov permite comparar los datos generados con una distribución $Exponencial(5)$.
El resultado obtenido presenta $p>0.05$, no se rechaza $H_0$, por lo que no se encuentra evidencia estadística suficiente para afirmar que los datos generados difieren de una distribución $Exponencial(5)$.

### Ejercicio 4
 **werbull($\alpha$,1)**

### Función de Densidad de Probabilidad (f.d.p.)
Dado $X \sim \text{Weibull}(\alpha, 1)$ para $x > 0$ y $\alpha > 0$:

$$f_X(x) = \alpha x^{\alpha - 1} e^{-x^\alpha}$$

### Función de Distribución Acumulada (F.D.A.)
Integramos la densidad desde el límite inferior $0$ hasta $x$:

$$F_X(x) = \int_{0}^{x} \alpha t^{\alpha - 1} e^{-t^\alpha} \, dt$$

Haciendo el cambio de variable $u = t^\alpha \implies du = \alpha t^{\alpha - 1} dt$:

$$F_X(x) = \int_{0}^{x^\alpha} e^{-u} \, du = \left[ -e^{-u} \right]_{0}^{x^\alpha} = -e^{-x^\alpha} - (-e^0)$$

$$F_X(x) = 1 - e^{-x^\alpha}$$

### Inversión (Despeje de $x$)
Igualamos la F.D.A. a una variable aleatoria uniforme $U \sim \text{Unif}(0,1)$:

$$U = 1 - e^{-x^\alpha}$$

$$e^{-x^\alpha} = 1 - U$$

$$\ln(e^{-x^\alpha}) = \ln(1 - U)$$

$$-x^\alpha = \ln(1 - U)$$

$$x^\alpha = -\ln(1 - U)$$

$$x = \left(-\ln(1 - U)\right)^{\frac{1}{\alpha}}$$

```r
geneweibull = function(nsim, alpha) {
  u = runif(nsim)
  x = (-log(u))^(1 / alpha)
  return(x)
}

nsim = 3000

alpha = 2
datos_simulados = geneweibull(nsim, alpha)

par(mfrow = c(1,2))
hist(datos_simulados, freq = FALSE, breaks = "FD", col = "purple",main = substitute(paste("Simulación Weibull(", alpha, ", 1) por Inversión"), list(alpha = alpha)))
curve(dweibull(x, shape = alpha, scale = 1), add = TRUE, col = "green", lwd = 3)
boxplot(datos_simulados, horizontal = TRUE,col = "purple")
```

#### Observaciones

* Graficamente se puede observar que la funcion con que contruimos con el metodo inverso esta generando los datos esperados con una distribucion werbull

### Ejercicio 5
**Gumbel(0,1)**

### Función de Densidad de Probabilidad (f.d.p.)
Dado $X \sim \text{Gumbel}(0, 1)$:

$$f_X(x) = e^{-x} e^{-e^{-x}}$$

### Función de Distribución Acumulada (F.D.A.)
Integramos la densidad desde $-\infty$ hasta $x$:

$$F_X(x) = \int_{-\infty}^{x} e^{-t} e^{-e^{-t}} \, dt$$

Haciendo el cambio de variable $u = e^{-t} \implies du = -e^{-t} dt$:
* Cuando $t \to -\infty \implies u \to \infty$
* Cuando $t = x \implies u = e^{-x}$

$$F_X(x) = \int_{\infty}^{e^{-x}} -e^{-u} \, du = \int_{e^{-x}}^{\infty} e^{-u} \, du = \left[ -e^{-u} \right]_{e^{-x}}^{\infty} = 0 - (-e^{-e^{-x}})$$

$$F_X(x) = e^{-e^{-x}}$$

### Inversión (Despeje de $x$)
Igualamos la F.D.A. a una variable aleatoria uniforme $U \sim \text{Unif}(0,1)$:

$$U = e^{-e^{-x}}$$

Aplicamos logaritmo natural ($\ln$) en ambos lados:

$$\ln(U) = -e^{-x}$$

Multiplicamos por $-1$:

$$-\ln(U) = e^{-x}$$

Aplicamos logaritmo natural nuevamente:

$$\ln(-\ln(U)) = -x$$

Despejamos $x$:

$$x = -\ln(-\ln(U))$$

```r
genergumbel = function(nsim) {
  u = runif(nsim)
  x = -log(-log(u))
  return(x)
}

nsim = 3000
datos_simulados = genergumbel(nsim)

dgumbel_n = function(x) {
  exp(-x) * exp(-exp(-x))
}

par(mfrow = c(1,2))
hist(datos_simulados, freq = FALSE, breaks = "FD", col = "purple",main = "Simulación Gumbel(0, 1) por Inversión")
curve(dgumbel_n(x), add = TRUE, col = "green", lwd = 3)
boxplot(datos_simulados,horizontal = TRUE,col = "purple")
```

#### Observaciones

* Graficamente se puede observar que la funcion con que contruimos con el metodo inverso esta generando los datos esperados con una distribucion gumbel

### Ejercicio 6
**Pareto($\alpha$)**

### Función de Densidad de Probabilidad (f.d.p.)
Dado $X \sim \text{Pareto}(\alpha, 1)$ para $x \ge 1$ y $\alpha > 0$:

$$f_X(x) = \frac{\alpha}{x^{\alpha + 1}}$$

### Función de Distribución Acumulada (F.D.A.)
Integramos la densidad desde el límite inferior $1$ hasta $x$:

$$F_X(x) = \int_{1}^{x} \frac{\alpha}{t^{\alpha + 1}} \, dt = \int_{1}^{x} \alpha t^{-\alpha - 1} \, dt$$

$$F_X(x) = \left[ -t^{-\alpha} \right]_{1}^{x} = -x^{-\alpha} - (-1^{-\alpha})$$

$$F_X(x) = 1 - x^{-\alpha}$$

### Inversión (Despeje de $x$)
Igualamos la F.D.A. a una variable aleatoria uniforme $U \sim \text{Unif}(0,1)$:

$$U = 1 - x^{-\alpha}$$

$$x^{-\alpha} = 1 - U$$

Elevamos ambos lados a la potencia $-\frac{1}{\alpha}$:

$$(x^{-\alpha})^{-\frac{1}{\alpha}} = (1 - U)^{-\frac{1}{\alpha}}$$

$$x = (1 - U)^{-\frac{1}{\alpha}} = \frac{1}{(1 - U)^{\frac{1}{\alpha}}}$$


```r
gen_pareto_inversion <- function(nsim, alpha) {
  u=runif(nsim)
  x=(1 - u)^(-1 / alpha)
  return(x)
}
dpareto_n = function(x, alpha) {
  ifelse(x >= 1, alpha / (x^(alpha + 1)), 0)
}

nsim = 3000
alpha = 3
datos_simulados = gen_pareto_inversion(nsim, alpha)

hist(datos_simulados, freq = FALSE, breaks = "FD", col = "purple",
     xlim = c(1, quantile(datos_simulados, 0.99)),
     main = substitute(paste("Simulación Pareto(", alpha, ", 1)")))

curve(dpareto_n(x, alpha), add = TRUE, col = "green", lwd = 3)
```

#### Observaciones

* Graficamente se puede observar que la funcion con que contruimos con el metodo inverso esta generando los datos esperados con una distribucion pareto

### Ejercicio 7

Sean $z_1, z_2, \dots, z_n$ variables aleatorias independientes e idénticamente distribuidas (**i.i.d.**) tales que:

$$z_i \sim \mathcal{N}(0,1), \quad \forall i = 1, \dots, n$$

Definimos la variable aleatoria $X$ como:

$$X = \sum_{i=1}^{n} z_i^2 \sim \mathcal{X}_n^2$$

donde $n \in \mathbb{Z}^+$ representa el tamaño de la muestra y $\bar{x}_n = \frac{1}{n}\sum_{i=1}^{n} z_i$ es la media muestral.

```r
library("car")

chi = function(n,nsim){
  matrix1 = matrix(rnorm(n * nsim),nrow = nsim,ncol = n)

  valores_x = rowSums(matrix1^2)

  par(mfrow = c(2,2))
  hist(valores_x,freq = FALSE,breaks = "FD",col = "cyan",main = "Hist X^2")
  curve(dchisq(x,df = n),add = TRUE,lwd = 3,col = "purple")
  qqPlot(valores_x)
  boxplot(valores_x,horizontal = TRUE,col = "cyan")
  acf(valores_x)

}

chi(3,2000)
```

#### Observaciones

* El ajuste entre el histograma empírico y la curva teórica (morada) confirma visualmente que la suma de variables normales al cuadrado sigue la distribución Chi-cuadrado esperada
* La curvatura pronunciada respecto a la línea azul recta demuestra claramente que los datos simulados no siguen una distribución Normal

* Muestra una marcada asimetría hacia la derecha con la mediana sesgada hacia la izquierda (cerca de $x = 2$) y una alta concentración de datos atípicos o extremos hacia la derecha (valores por encima de $8$), lo cual es característico del comportamiento de la distribución Chi-cuadrado

* a ausencia de barras significativas fuera de las bandas de confianza de decisión (a excepción del rezago $0$) valida la independencia estadística entre las simulaciones generadas

### Ejercicio 8
Sean $z$ e $y$ dos variables aleatorias **independientes** tales que:

$$z \sim \mathcal{N}(0,1)$$
$$y \sim \chi_n^2$$

Definimos la variable aleatoria $X$ como:

$$X = \frac{z}{\sqrt{y / n}}$$

Entonces, la variable $X$ se distribuye como una **t de Student con $n$ grados de libertad**:

$$X \sim t_n$$

```r
t = function(n,nsim){
  z = rnorm(nsim)
  y = rchisq(nsim,df = n)

  X = z/sqrt(y/n)

  par(mfrow = c(2,2))
  hist(X,freq = FALSE,breaks = "FD",col = "purple",main = "hist X")
  curve(dt(x,df = n),add = TRUE,lwd = 3,col = "cyan")
  qqPlot(X)
  boxplot(X,horizontal = TRUE,col = "purple",main = "boxplot X")
  acf(X)
}

t(3,1000)
```

#### Observaciones

* Las barras moradas siguen de forma precisa la curva de densidad cian $t(df = 3)$. Muestra una distribución acampanada, centrada en $0$ y con colas más pesadas que las de una Normal Estándar, lo que confirma que la relación $X = \frac{z}{\sqrt{y/n}}$ genera la distribución $t$ de Student esperada

* La forma de "S" pronunciada (los puntos se desvían de la línea recta azul tanto en la cola izquierda como en la derecha) demuestra que los datos tienen colas más pesadas que una Normal Estándar. Esto es totalmente correcto, pues la distribución $t_n$ asigna mayor probabilidad a valores extremos que la Normal

* El gráfico es simétrico con respecto a $0$ (su mediana está centrada), pero exhibe una cantidad considerable de valores atípicos en ambas direcciones (valores $< -3$ y $> 3$). Esto respalda de nuevo la presencia de colas pesadas debidas a los pocos grados de libertad ($n = 3$).

### Ejercicio 9

Sean $y_1$ e $y_2$ dos variables aleatorias **independientes** tales que:

$$y_1 \sim \chi_{n_1}^2 \quad \text{y} \quad y_2 \sim \chi_{n_2}^2$$

Definimos la variable aleatoria $X$ como el cociente de ambas variables divididas por sus respectivos grados de libertad:

$$X = \frac{y_1 / n_1}{y_2 / n_2}$$

Entonces, $X$ sigue una **distribución F de Snedecor** con $n_1$ y $n_2$ grados de libertad:

$$X \sim F_{n_1, n_2}$$

```r
Fi = function(n1,n2,nsim){
  y1 = rchisq(nsim,df = n1)
  y2 = rchisq(nsim,df = n2)

  X = (y1/n1)/(y2/n2)

  par(mfrow = c(2,2))
  hist(X,freq = FALSE,breaks = "FD",col = "yellow",main = "hist X")
  curve(df(x,df1 = n1,df2 = n2),add = TRUE,lwd = 3,col = "red")
  qqPlot(X)
  boxplot(X,horizontal = TRUE,col = "yellow",main = "boxplot X")
  acf(X)

}

Fi(4,15,1000)
```

#### Observaciones

* Histograma (Arriba a la izquierda): Las barras amarillas encajan perfectamente con la línea roja. Esto nos dice de forma súper visual que la fórmula funciona: al dividir las dos Chi-cuadrado ajustadas, los datos caen exactamente donde la teoría de la distribución F dice que deberían caer. Además, no hay valores negativos porque estás dividiendo números positivos.

* La línea de puntos se dispara hacia arriba al final y no sigue la línea recta azul. Esto es una prueba clarísima de que estos datos para nada son una Normal. Se curva así porque la distribución F tiene una "cola" muy larga hacia la derecha.

* La caja está apretada cerca del 1 y tiene un montón de puntos sueltos hacia la derecha (que llegan hasta el 8 o 9). Esto te muestra que la mayoría de los resultados son números pequeños, pero de vez en cuando la división da un salto y genera valores bastante altos

* Todas las barras (menos la primera, que siempre es 1) están dentro de las líneas punteadas azules. En español sencillo: cada dato que simulaste es completamente independiente del anterior, no hay ningún patrón ni "trampa" en cómo R generó los números aleatorios.

## Algoritmo Box-Muller - Ejercicio 10

```r
# algoritmo Box-Muller


box_muller = function(nsim){
  x = rep(0,nsim)
  y = rep(0,nsim)

  for(i in 1:nsim){

    u1 = runif(nsim)
    u2 = runif(nsim)

    R = sqrt(-2*log(u1))
    theta = 2*pi*u2

    x[i] = R*cos(theta)
    y[i] = R*sin(theta)
  }
  return(list(x = x,y = y))
}

X_salida = box_muller(1000)$x

Y_salida = box_muller(1000)$y
```

## Marsaglia - Ejercicio 11

El **método de Marsaglia** (o transformación de *Marsaglia-Bray*) es una modificación del algoritmo de Box-Muller diseñada para evitar el cálculo de funciones trigonométricas ($\cos$ y $\sin$), las cuales son computacionalmente costosas.

En lugar de generar un ángulo $\theta$ directamente con $2\pi U_2$, Marsaglia utiliza un método de **aceptación y rechazo** en el círculo unitario:

1. Genera dos variables uniformes independientes $V_1, V_2 \sim \text{Unif}(-1, 1)$.
2. Calcula la distancia al cuadrado desde el origen: $S = V_1^2 + V_2^2$.
3. Si $S \ge 1$ o $S = 0$, se descarta el punto y se vuelve a intentar (aceptamos solo puntos dentro del círculo unitario de radio 1).
4. Dentro del círculo unitario, los cocientes $\frac{V_1}{\sqrt{S}}$ y $\frac{V_2}{\sqrt{S}}$ representan exactamente el coseno y seno de un ángulo uniforme.

Sustituyendo estas relaciones en la fórmula original de Box-Muller, se obtienen las expresiones finales:

$$Z_1 = V_1 \sqrt{\frac{-2 \ln(S)}{S}}$$

$$Z_2 = V_2 \sqrt{\frac{-2 \ln(S)}{S}}$$

```r
masaglia = function(nsim){
  X = rep(0,nsim)
  Y = rep(0,nsim)

  for(i in 1:nsim){
    w = 2
    while ( w>1 || w==0){
      u1 = runif(1)
      u2 = runif(1)

      v1 = 2*u1-1
      v2 = 2*u2-1

      w = v1^2 + v2^2
    }
    c = sqrt((-2 * log(w)) / w)

    X[i] = c * v1
    Y[i] = c * v2
  }

  return(list(x = X, y = Y))
}

X = masaglia(2000)$x
Y = masaglia(2000)$y
```

# Box-Muller vs Marsaglia

El método de Box-Muller garantiza una transformación analítica biyectiva sin pérdida de muestras, pero su rendimiento está acotado por la latencia del procesador al evaluar aproximaciones polinomiales para senos y cosenos. Por su parte, la variante de Marsaglia sustituye el costo computacional de las funciones trigonométricas por una prueba de aceptación y rechazo geométrica, reduciendo los ciclos de reloj totales a pesar de la tasa de descarte de variables para un coste menor computacionalmente es mejor Marsaglia ya teniendo los tiempos comparados anterior mente sigue siendo demasiado mejor marsaglia

Derivando respecto a $x$:$$\frac{d}{dx}\ln k(x) = \frac{\alpha - 1}{x} - \frac{\beta - 1}{1 - x} = 0$$

En el procedimiento se utilizará la constante:

El algoritmo consiste en generar una variable $T$ de la distribución Laplace y una variable $U\sim U(0,1)$, aceptando el valor $T$ cuando se cumple:

$$U\leq\frac{f(T)}{c,g(T)}$$

```r
count = 0

g=function(x){
   return(0.5*exp(-abs(x)))
}

f=function(x){
    return((1/sqrt(2*pi))*exp(-x^2/2))
}

fg=function(x){
    return(f(x)/g(x))
}


rnormal = function(){
  c=1.3154
  while(TRUE){
    count <<- count + 1
    u = runif(1)
    T = rlaplace(1)
    if(u <= f(T)/(c *g(T))){
      return(T)
    }
  }
}

rgnormal = function(nsim){
  X = c()
  count = 0
  for(i in 1:nsim){
    X[i] = rnormal()
  }
  return(X)
}

Nsim = 5000
count = 0
datos2 = rgnormal(Nsim)
```
